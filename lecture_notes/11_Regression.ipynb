## Linear Regression

Linear regression is a statistical method used to model the relationship between a **response variable** $y$ and one or more **predictor variables** $x_1, x_2, \dots, x_p$. The key idea is that the model is **linear in the parameters**.

### The Linear Model

A linear regression model has the form

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_p x_p + \varepsilon
$$

where:

- $\beta_0, \dots, \beta_p$ are **unknown parameters**,
- $\varepsilon$ is a random error term with mean $0$.

The word "linear" means **linear in the coefficients** $\beta_j$, not necessarily linear in the variables $x_j$.

### Examples
#### Linear (still a linear regression model)
$$
y = \beta_0 + \beta_1 x + \beta_2 x^2 + \varepsilon
$$
Even though $x^2$ is nonlinear in $x$, the model is linear in $\beta$.

#### Not linear in parameters
$$
y = \beta_0 + \beta_1 x^{\beta_2} + \varepsilon
$$
Here $\beta_2$ appears inside a nonlinear function, so the model is **nonlinear regression**.

### Matrix Form
It is convenient to write linear regression in matrix notation:
$$
\mathbf{y} = X\beta + \varepsilon
$$
where
- $\mathbf{y}$ is an $n \times 1$ vector of responses,
- $X$ is an $n \times (p+1)$ **design matrix**,
- $\beta$ is a $(p+1)$ vector of parameters.

### Least Squares Estimation

The most common way to estimate $\beta$ is **ordinary least squares (OLS)**:

$$
\hat{\beta} = \arg\min_\beta \| \mathbf{y} - X\beta \|^2
$$

If $X^\top X$ is invertible, the solution has the closed form

$$
\hat{\beta} = (X^\top X)^{-1} X^\top \mathbf{y}
$$

### Connection to Gaussian Errors

Suppose the errors are independent and Gaussian:

$$
\varepsilon_i \overset{\text{i.i.d.}}{\sim} \mathcal{N}(0, \sigma^2)
$$

Then

$$
y_i \mid x_i \sim \mathcal{N}(x_i^\top \beta, \sigma^2)
$$

and the likelihood is

$$
L(\beta, \sigma^2) \propto (\sigma^2)^{-n/2}\exp\!\left(-\frac{1}{2\sigma^2}\sum_{i=1}^n (y_i - x_i^\top\beta)^2\right)
$$

Maximizing this likelihood with respect to $\beta$ is equivalent to minimizing the residual sum of squares

$$
\sum_{i=1}^n (y_i - x_i^\top\beta)^2
$$

so under independent Gaussian errors, OLS is exactly the **maximum likelihood estimator (MLE)** for $\beta$.

### Interpretation

- Each coefficient $\beta_j$ represents the **expected change in $y$ per unit change in $x_j$**, holding all other predictors fixed.
- The fitted values are $\hat{y}_i = x_i^\top \hat{\beta}$.
- The residuals are $e_i = y_i - \hat{y}_i$.

### Why Linear Regression?

- Simple and interpretable
- Fast to compute
- Closed-form solution (when assumptions hold)
- Forms a foundation for many advanced models
